## Limites de la malla peatonal y vial

```python
import geopandas as gpd
import matplotlib.pyplot as plt
from shapely.geometry import box

# 1. Cargar las mallas y reproyectar a EPSG:3116 (Metros - Bogotá/Colombia)
TARGET_CRS = "EPSG:3116"
gdf_red_vial = gpd.read_file("Data/raw/Malla_Vial_Integral_Bogota_D_C (3) (1).geojson").to_crs("EPSG:3116")
gdf_red_peatonal = gpd.read_file("Data/raw/red_peatonal_conexion (1).gpkg").to_crs("EPSG:3116")

# 2. Calcular el Bounding Box de la INTERSECCIÓN directamente en metros (EPSG:3116)
bbox_vial = gdf_red_vial.total_bounds
bbox_peatonal = gdf_red_peatonal.total_bounds

xmin_intersec = max(bbox_vial[0], bbox_peatonal[0])
ymin_intersec = max(bbox_vial[1], bbox_peatonal[1])
xmax_intersec = min(bbox_vial[2], bbox_peatonal[2])
ymax_intersec = min(bbox_vial[3], bbox_peatonal[3])

# Crear el polígono del área común directamente en EPSG:3116
area_comun_poly = box(xmin_intersec, ymin_intersec, xmax_intersec, ymax_intersec)
gdf_box = gpd.GeoDataFrame({'geometry': [area_comun_poly]}, crs=TARGET_CRS)

# 3. Graficar con las escalas unificadas en metros
fig, ax = plt.subplots(figsize=(10, 10))

# Graficar malla vial
gdf_red_vial.plot(
    ax=ax, 
    color='#aaaaaa', 
    linewidth=0.5, 
    alpha=0.6, 
    label='Malla Vial Contexto'
)

# Graficar el área común ajustada correctamente en EPSG:3116
gdf_box.plot(
    ax=ax, 
    facecolor='red', 
    alpha=0.35, 
    edgecolor='red', 
    linewidth=2, 
    label='Área Común (BBOX Intersección)'
)

# 4. Ajustar el enfoque (zoom) al área de estudio
ax.set_xlim(xmin_intersec - 2000, xmax_intersec + 2000)
ax.set_ylim(ymin_intersec - 2000, ymax_intersec + 2000)

plt.title("Área Común de Análisis - Sistema Proyectado EPSG:3116 (Metros)", fontsize=12, fontweight='bold')
plt.xlabel("Este / X (m)")
plt.ylabel("Norte / Y (m)")
plt.grid(True, linestyle='--', alpha=0.4)
plt.legend(loc='upper left')

plt.tight_layout()
plt.show()
```

```python
from shapely.ops import unary_union

# Se genera un polígono de "cobertura" por cada malla: un buffer pequeño alrededor de cada
# línea, disuelto en un solo polígono. El buffer es necesario porque unir líneas directamente
# no genera área — se necesita un polígono para poder intersectar coberturas.
BUFFER_COBERTURA = 100  # metros; representa el radio de influencia de una vía/anden

cobertura_vial = unary_union(gdf_red_vial.geometry.buffer(BUFFER_COBERTURA))
cobertura_peatonal = unary_union(gdf_red_peatonal.geometry.buffer(BUFFER_COBERTURA))

# La intersección real respeta la forma irregular de ambas mallas, a diferencia del bbox.
area_comun_real = cobertura_vial.intersection(cobertura_peatonal)

gdf_area_comun = gpd.GeoDataFrame({'geometry': [area_comun_real]}, crs=TARGET_CRS)
```

```python
# Comparación honesta: cuánta área "de más" estaba incluyendo el bbox vs la intersección real.
area_bbox_km2 = gdf_box.geometry.area.iloc[0] / 1e6
area_real_km2 = gdf_area_comun.geometry.area.iloc[0] / 1e6
print(f"Área bbox: {area_bbox_km2:.2f} km2")
print(f"Área intersección real: {area_real_km2:.2f} km2")
print(f"Sobreestimación del bbox: {(area_bbox_km2/area_real_km2 - 1)*100:.1f}%")
```

```python
import matplotlib.pyplot as plt

# Se grafican ambas mallas de contexto, el bbox anterior (para comparar) y el área común real,
# todo en el mismo eje para visualizar cuánta zona "de más" incluía el bbox.
fig, ax = plt.subplots(figsize=(12, 12))

# Mallas de contexto en gris tenue, solo como referencia espacial de fondo.
gdf_red_vial.plot(
    ax=ax,
    color='#aaaaaa',
    linewidth=0.5,
    alpha=0.5,
    label='Malla Vial (contexto)'
)
gdf_red_peatonal.plot(
    ax=ax,
    color='#888888',
    linewidth=0.5,
    alpha=0.5,
    label='Malla Peatonal (contexto)'
)

# Bbox anterior en azul punteado, para dejar visible el error de sobreestimación.
gdf_box.boundary.plot(
    ax=ax,
    color='blue',
    linewidth=2,
    linestyle='--',
    label='Área común (bbox - sobreestimada)'
)

# Área común real (intersección de coberturas) en rojo sólido, esta es la que se usa de aquí en adelante.
gdf_area_comun.plot(
    ax=ax,
    facecolor='red',
    alpha=0.35,
    edgecolor='red',
    linewidth=2,
    label='Área común real (intersección)'
)

ax.set_title("Comparación: Área común por Bounding Box vs Intersección real de coberturas", fontsize=13)
ax.set_xlabel("Este (m) - EPSG:3116")
ax.set_ylabel("Norte (m) - EPSG:3116")
ax.legend(loc='lower left')
ax.set_aspect('equal')  # Obligatorio en CRS proyectado: sin esto el mapa se deforma visualmente.
plt.tight_layout()
plt.savefig("comparacion_area_comun.png", dpi=150)
plt.show()
```

### **Diccionario de Variables: Columna `FUNCIONAMIENTO`**

**Definición del Dominio:** Posibles combinaciones entre las diferentes estrategias de control a nivel de controlador local y a nivel de central, que permiten gestionar de manera más eficiente las diferentes operaciones de la intersección semafórica *(Fuente: Documento Institucional GIG-06-FR-02 V.2)*.

| Código | Estrategia en Central | Estrategia en Controlador | Descripción de la Operación Semafórica | Frecuencia |
| :---: | :--- | :--- | :--- | :---: |
| **BT-TF** | Basado en Tiempo (`BT`) | Tiempos Fijos (`TF`) | Selección de planes por horario desde la central + ejecución fija preestablecida en el controlador. | 508 |
| **BT-TA** | Basado en Tiempo (`BT`) | Tráfico Actuado (`TA`) | Selección de planes por horario desde la central + optimización por algoritmo y detectores en el sitio. | 328 |
| **CA-TA** | Central Adaptativo (`CA`) | Tráfico Actuado (`TA`) | Ajuste en tiempo real de verdes/fases por detectores en central + optimización variable en el controlador. | 306 |
| **RP-TF** | Responsivo (`RP`) | Tiempos Fijos (`TF`) | Selección responsiva de planes enviada desde central según detectores + ejecución fija en el sitio. | 281 |
| **R-TF** | Responsivo (`R`) | Tiempos Fijos (`TF`) | Cambio de planes activado por demanda de tráfico + regulación constante en el controlador. | 107 |
| **R-TA** | Responsivo (`R`) | Tráfico Actuado (`TA`) | Cambio de planes activado por demanda de tráfico + ajuste por algoritmo/detectores en el sitio. | 103 |
| **RP-TA** | Responsivo (`RP`) | Tráfico Actuado (`TA`) | Selección responsiva desde central + optimización variable por detectores en el sitio. | 100 |
| **CA-TF** | Central Adaptativo (`CA`) | Tiempos Fijos (`TF`) | Control adaptativo dinámico en central + tiempos preestablecidos locales en el controlador. | 12 |

---

#### **Nomenclatura de Estrategias de Control:**

* **Control a Nivel de Central (Primeras Siglas):**
  * `BT` (**Basado en el Tiempo**): Mecanismo de selección de planes de señales activados en función de un horario o día específico.
  * `CA` (**Central Adaptativo**): Modo de control que ajusta los planes de señales basándose en las condiciones del tránsito registradas por los detectores (ajuste de tiempos de luz verde y otras acciones).
  * `RP` / `R` (**Responsivo**): Mecanismo de selección que ajusta los planes enviando órdenes de cambio a un plan determinado según las condiciones del tránsito registradas por detectores.

* **Control a Nivel de Controlador / Local (Segundas Siglas):**
  * `TA` (**Tráfico Actuado**): Control basado en programación variable accazada por un algoritmo en el controlador, optimizando la operación según las variables del tránsito detectadas en el sitio.
  * `TF` (**Tiempos Fijos**): Programación fija de planes que cambian de forma preestablecida en el controlador de manera constante y predecible, sin depender del tráfico en tiempo real.

```python
gdf_red_peatonal_clip = gpd.clip(gdf_red_peatonal, gdf_area_comun)

# Verificación obligatoria: clip puede generar geometrías vacías o de tipo distinto
# (ej. Point si una línea solo toca el borde en un punto). Se descartan antes de segmentar.
gdf_red_peatonal_clip = gdf_red_peatonal_clip[
    (~gdf_red_peatonal_clip.geometry.is_empty) &
    (gdf_red_peatonal_clip.geometry.notna()) &
    (gdf_red_peatonal_clip.geom_type.isin(["LineString", "MultiLineString"]))
].copy()

print(f"Segmentos originales: {len(gdf_red_peatonal)} | Tras clip a área común: {len(gdf_red_peatonal_clip)}")
```

```python
import numpy as np
from shapely.geometry import LineString

# Verificación obligatoria ANTES de segmentar: el CRS de entrada debe ser exactamente el
# esperado. Esto no reproyecta -- reproyectar aquí escondería el error en vez de exponerlo.
# Si esto falla, el problema está upstream (en la carga/reproyección de gdf_red_peatonal),
# no en la segmentación, y hay que corregirlo ahí, no parchear en este punto.
assert gdf_red_peatonal_clip.crs is not None, "gdf_red_peatonal_clip no tiene CRS definido"
assert gdf_red_peatonal_clip.crs.to_epsg() == 3116, (
    f"CRS inesperado en gdf_red_peatonal_clip: {gdf_red_peatonal_clip.crs}. "
    f"Se esperaba EPSG:3116. Revisar la reproyección upstream antes de segmentar."
)

def segmentar_linea(linea, longitud_segmento=200):
    """
    Corta una LineString en tramos de longitud fija con interpolate().
    El último tramo queda más corto que 200 m — se conserva así para no perder
    malla residual, y se corrige después vía normalización por longitud real.
    """
    distancias = list(np.arange(0, linea.length, longitud_segmento)) + [linea.length]
    puntos = [linea.interpolate(d) for d in distancias]
    return [LineString([puntos[i], puntos[i + 1]]) for i in range(len(puntos) - 1)]

# explode() descompone MULTILINESTRING en LINESTRING individuales; interpolate() no opera
# sobre multigeometrías, por lo que este paso es obligatorio antes de segmentar.
gdf_red_exp = gdf_red_peatonal_clip.explode(index_parts=False).reset_index(drop=True)

filas_segmentos = []
for idx, fila in gdf_red_exp.iterrows():
    for i, seg in enumerate(segmentar_linea(fila.geometry, 200)):
        registro = fila.drop("geometry").to_dict()
        registro["geometry"] = seg
        registro["id_segmento"] = f"{idx}_{i}"
        filas_segmentos.append(registro)

# Se hereda el CRS directamente de la fuente (gdf_red_exp.crs) en vez de hardcodear
# TARGET_CRS a mano -- así, si la fuente cambiara de CRS por error, el GeoDataFrame resultante
# lo reflejaría correctamente en vez de mentir con una etiqueta fija que ya no corresponde.
gdf_segmentos = gpd.GeoDataFrame(filas_segmentos, geometry="geometry", crs=gdf_red_exp.crs)
gdf_segmentos["long_m"] = gdf_segmentos.geometry.length

# Chequeo de sanidad inmediato, no esperar al Nivel 2 completo para detectar un CRS roto.
# Un segmento de 200m real en grados daría long_m ~0.0018, no ~60-200.
assert gdf_segmentos["long_m"].max() <= 200 * 1.01, (
    "long_m excede el máximo esperado -- revisar si el CRS de origen realmente era metros."
)
print(f"Segmentos generados: {len(gdf_segmentos)} | long_m rango: "
      f"[{gdf_segmentos['long_m'].min():.2f}, {gdf_segmentos['long_m'].max():.2f}]")
```

```python
# Antes de unir semáforos, se recorta también esa capa al área común real, por consistencia
# con la red peatonal — evita traer semáforos fuera del universo espacial del modelo.
gdf_sem_clip = gpd.clip(gdf_sem, gdf_area_comun)
gdf_sem_clip = gdf_sem_clip[gdf_sem_clip.geometry.notna() & ~gdf_sem_clip.geometry.is_empty].copy()

# Buffer de 50 m como zona de influencia por segmento.
buffers = gdf_segmentos.copy()
buffers["geometry"] = buffers.geometry.buffer(50)

# predicate="intersects": un semáforo justo en el borde del buffer también cuenta.
# how="left" conserva segmentos sin ningún semáforo cercano (NaN, no se descartan).
join_sem = gpd.sjoin(buffers, gdf_sem_clip, how="left", predicate="intersects")
```

```python
# --- Indicador base: conteo y densidad de semáforos por segmento ---
n_semaforos = join_sem.groupby("id_segmento").size().rename("n_semaforos")
gdf_segmentos = gdf_segmentos.merge(n_semaforos, on="id_segmento", how="left")
gdf_segmentos["n_semaforos"] = gdf_segmentos["n_semaforos"].fillna(0)

# Densidad por km, no conteo bruto — necesario porque el último tramo de cada vía no mide 200 m.
gdf_segmentos["densidad_semaforos_km"] = (
    gdf_segmentos["n_semaforos"] / (gdf_segmentos["long_m"] / 1000)
)

# --- Indicador de accesibilidad peatonal/sonora (igual que antes) ---
tiene_peatonal = join_sem["GRUPOSPEATONALES"].notna() & (join_sem["GRUPOSPEATONALES"] != "")
n_peatonal = join_sem[tiene_peatonal].groupby("id_segmento").size().rename("n_sem_peatonal")
gdf_segmentos = gdf_segmentos.merge(n_peatonal, on="id_segmento", how="left")
gdf_segmentos["n_sem_peatonal"] = gdf_segmentos["n_sem_peatonal"].fillna(0)

tiene_sonoro = join_sem["GRUPOSSONOROS"].notna() & (join_sem["GRUPOSSONOROS"] != "")
n_sonoro = join_sem[tiene_sonoro].groupby("id_segmento").size().rename("n_sem_sonoro")
gdf_segmentos = gdf_segmentos.merge(n_sonoro, on="id_segmento", how="left")
gdf_segmentos["n_sem_sonoro"] = gdf_segmentos["n_sem_sonoro"].fillna(0)

# --- Indicador nuevo: semáforos con falla, como variable de riesgo (punto 1 de la sesión anterior) ---
# AJUSTAR los strings tras revisar value_counts real de FUNCIONAMIENTO.
valores_falla = ["Dañado", "Intermitente", "Fuera de Servicio"]
join_sem["sem_con_falla"] = join_sem["FUNCIONAMIENTO"].isin(valores_falla)

n_falla = join_sem[join_sem["sem_con_falla"]].groupby("id_segmento").size().rename("n_sem_con_falla")
gdf_segmentos = gdf_segmentos.merge(n_falla, on="id_segmento", how="left")
gdf_segmentos["n_sem_con_falla"] = gdf_segmentos["n_sem_con_falla"].fillna(0)

# Proporción de falla sobre el total de semáforos del segmento — más informativo que el conteo bruto
# porque aísla si la falla es sistemática en esa zona o solo un caso aislado entre varios semáforos sanos.
# where evita división por cero en segmentos sin semáforos (queda NaN en vez de inf).
gdf_segmentos["pct_sem_con_falla"] = (
    gdf_segmentos["n_sem_con_falla"] / gdf_segmentos["n_semaforos"].where(gdf_segmentos["n_semaforos"] > 0)
)
```

```python
# La longitud de cada segmento debe estar en [algo pequeño, 200]. Si hay valores muy por
# encima de 200, la función de segmentación no está cortando correctamente.
print("Estadísticas de long_m:")
print(gdf_segmentos["long_m"].describe())

# Cualquier segmento con longitud 0 o negativa es un error de la función interpolate(),
# probablemente por una LineString degenerada (dos puntos duplicados).
segmentos_rotos = gdf_segmentos[gdf_segmentos["long_m"] <= 0]
print(f"Segmentos con longitud <= 0: {len(segmentos_rotos)}")

# El área común, en km2, debe ser un número razonable para Bogotá urbana (~300-400 km2
# aprox de zona urbana consolidada). Si te da 3000 km2 o 3 km2, algo se rompió en la intersección.
print(f"Área común: {gdf_area_comun.geometry.area.iloc[0] / 1e6:.2f} km2")

# Conteo de semáforos por segmento no debería tener outliers absurdos (ej. 200 semáforos
# en un segmento de 200m es geométricamente imposible salvo error de buffer/CRS).
print("Estadísticas de n_semaforos por segmento:")
print(gdf_segmentos["n_semaforos"].describe())
print(f"Máximo n_semaforos en un solo segmento: {gdf_segmentos['n_semaforos'].max()}")
```
