# EDA — Siniestros Viales de Bogotá D.C. (Unión de fuentes)

Este cuaderno une y analiza dos fuentes de datos abiertos sobre siniestralidad vial en Bogotá D.C.:

1. **`DATO_ABIERTOS_siniestros_viales_consolidados_bogota_dc (3).xlsx`** — archivo oficial "Datos Abiertos" con 5 hojas relacionales: `SINIESTROS`, `ACTOR_VIAL`, `VEHICULOS`, `HIPOTESIS` y `DICCIONARIO` (codificación de valores).
2. **`historico_siniestros_bogota_d_c_-.csv`** — histórico consolidado de siniestros con coordenadas geográficas (`LATITUD`, `LONGITUD`).

Ambas fuentes comparten la llave **`CODIGO_ACCIDENTE`**, que identifica un siniestro. Se usa para:
- Unir el histórico georreferenciado (CSV) con el detalle oficial del siniestro (hoja `SINIESTROS`).
- Enriquecer cada siniestro con información agregada de actores viales, vehículos involucrados y causas (hipótesis), agregadas desde las otras hojas del Excel.

Estructura del cuaderno:

1. Diccionario de variables (CSV y cada hoja del Excel)
2. Carga de datos y entendimiento inicial de actor vial
3. Preparación y unión de los datasets por `CODIGO_ACCIDENTE`; filtrado a siniestros con peatón
4. EDA general del dataset filtrado a peatones
5. Análisis temporal
6. Análisis bivariado: estado del peatón, clase, causas, localidad y multicolinealidad
7. Mapas de zonas con siniestros que involucran peatones
8. Resumen, hallazgos clave y conclusiones
9. Salida final (exportación a GeoPackage)

**Nota de nomenclatura (2026-09-17):** a partir de la sección 3.4, `df` deja de ser el dataset a nivel de accidente (`df_total`, con la variable `GRAVEDAD`: SOLO DAÑOS / CON HERIDOS / CON MUERTOS) y pasa a ser el dataset a nivel de peatón individual (con la variable `ESTADO`: ILESO / HERIDO / MUERTO). Todo el análisis de la sección 6 en adelante opera sobre `ESTADO`, no sobre `GRAVEDAD` — los encabezados de esta versión ya se ajustaron para reflejarlo. Si el objetivo del capítulo de metodología es la severidad del *accidente* (`GRAVEDAD`), ese análisis bivariado está pendiente de rehacerse sobre `df_total`, no sobre este `df`.

## 3. Preparación y unión de los datasets

### 3.1 Función de decodificación

Se construye un diccionario Python `{{(HOJA, CAMPO): {{codigo: descripcion}}}}` a partir de la hoja `DICCIONARIO`, para traducir los códigos numéricos de `SINIESTROS`, `VEHICULOS` e `HIPOTESIS` a texto legible.

```python
def construir_mapa_diccionario(dicc):
    mapa = {}
    for _, row in dicc.iterrows():
        clave = (row['HOJA'], row['CAMPO'])
        mapa.setdefault(clave, {})[row['CODIGO']] = row['DESCRIPCION']
    return mapa

mapa_dicc = construir_mapa_diccionario(diccionario)

def decodificar(df, hoja, campo, nueva_col=None):
    nueva_col = nueva_col or f"{campo}_DESC"
    mapa = mapa_dicc.get((hoja, campo), {})
    df[nueva_col] = df[campo].map(mapa)
    return df

# Decodificar SINIESTROS
siniestros = decodificar(siniestros, 'SINIESTROS', 'GRAVEDAD')
siniestros = decodificar(siniestros, 'SINIESTROS', 'CLASE')
siniestros = decodificar(siniestros, 'SINIESTROS', 'CHOQUE')
siniestros = decodificar(siniestros, 'SINIESTROS', 'OBJETO_FIJO')
siniestros = decodificar(siniestros, 'SINIESTROS', 'CODIGO_LOCALIDAD', 'LOCALIDAD_DESC')
siniestros = decodificar(siniestros, 'SINIESTROS', 'DISENO_LUGAR')

# Decodificar VEHICULOS
vehiculos = decodificar(vehiculos, 'VEHICULOS', 'CLASE')
vehiculos = decodificar(vehiculos, 'VEHICULOS', 'SERVICIO')
vehiculos = decodificar(vehiculos, 'VEHICULOS', 'MODALIDAD')

# Decodificar HIPOTESIS
hipotesis = decodificar(hipotesis, 'HIPOTESIS', 'CODIGO_CAUSA')

siniestros[['GRAVEDAD','GRAVEDAD_DESC','CLASE','CLASE_DESC','DISENO_LUGAR','DISENO_LUGAR_DESC']].head()
```

### 3.4 Enriquecimiento con actores, vehículos y causas

Se agregan, por `CODIGO_ACCIDENTE`, resúmenes de las otras tres hojas y se incorporan al dataset unido:
- **Actores viales**: número de personas involucradas, número de muertos/heridos, sexo predominante.
- **Vehículos**: número de vehículos involucrados, clase de vehículo más frecuente.
- **Hipótesis**: causa principal reportada (la primera registrada por siniestro).

Para el EDA de siniestralidad **peatonal** el criterio correcto es analizar solo los accidentes donde estuvo involucrado un peaton.
1. Filtrar `ACTOR_VIAL` a `CONDICION == 'PEATON'`.
2. Calcular las métricas (conteo, muertos, heridos, sexo) **solo** sobre esos registros.
3. Unir esa agregación al dataset general con un **inner join** por `CODIGO_ACCIDENTE`, de modo que el dataset final conserve únicamente los siniestros en los que participó **al menos un peatón** — y no la totalidad de los siniestros.


```python
df = df_total.copy()
causa_principal = (
    hipotesis
    .sort_values('CODIGO_ACCIDENTE')
    .drop_duplicates('CODIGO_ACCIDENTE')
    .set_index('CODIGO_ACCIDENTE')['CODIGO_CAUSA_DESC']
    .rename('CAUSA_PRINCIPAL')
)

# 4. INFORMACIÓN INDIVIDUAL DE LOS PEATONES

peatones = actor_vial.loc[
    actor_vial['CONDICION'].eq('PEATON')
].copy()

# Edad numérica
peatones['EDAD'] = pd.to_numeric(
    peatones['EDAD'],
    errors='coerce'
)


# ============================================================
# 5. INFORMACIÓN DEL SINIESTRO
#    Se agrega a cada peatón
# ============================================================
info_siniestro = (
    siniestros[
        [
            'CODIGO_ACCIDENTE',
            'CODIGO_LOCALIDAD',
            'DISENO_LUGAR',
            'DIRECCION',
            'HORA',
            'CLASE',
        ]
    ]
    .drop_duplicates('CODIGO_ACCIDENTE')
    .set_index('CODIGO_ACCIDENTE')
)
peatones = (
    peatones
    .join(causa_principal, on='CODIGO_ACCIDENTE')
    .join(info_siniestro, on='CODIGO_ACCIDENTE')
)

info_csv = (
    df_total[
        [
            'CODIGO_ACCIDENTE',
            'LATITUD',
            'LONGITUD',
            'DISENO_LUGAR_DESC',
            'FECHA_OCURRENCIA_ACC',
            'ANO_OCURRENCIA_ACC',
            'FECHA_HORA_ACC'
        ]
    ]
    .drop_duplicates('CODIGO_ACCIDENTE')
    .set_index('CODIGO_ACCIDENTE')
)
peatones = (
    peatones
    .join(info_csv, on='CODIGO_ACCIDENTE'))

# ============================================================
# 6. ESTADÍSTICAS GENERALES
# ============================================================

n_general = df['CODIGO_ACCIDENTE'].nunique()

gravedad_general = (
    df[['CODIGO_ACCIDENTE', 'GRAVEDAD']]
    .drop_duplicates('CODIGO_ACCIDENTE')
    ['GRAVEDAD']
    .value_counts(normalize=True)
    .reindex(
        ['SOLO DANOS', 'CON HERIDOS', 'CON MUERTOS'],
        fill_value=0
    )
)


# ============================================================
# 7. DATASET FINAL A NIVEL DE PEATÓN
# ============================================================

df = peatones.copy()

n_peatonal = df['CODIGO_ACCIDENTE'].nunique()

print(
    f"Siniestros en el dataset general (todos los actores): "
    f"{n_general:,}"
)

print(
    f"Siniestros con al menos un peatón involucrado: "
    f"{n_peatonal:,}"
)

print(
    f"Porcentaje del total: "
    f"{n_peatonal / n_general * 100:.1f}%"
)

df.head()
```

### 6.1 Diseño del lugar y tipo de choque

```python
def miles(x, _=None):
    return f'{x:,.0f}'.replace(',', '.')

def pct_es(p):
    """Porcentaje con coma decimal; p en proporción (0-1)."""
    return f'{100 * p:.1f}'.replace('.', ',')

def _clave(s):
    s = unicodedata.normalize('NFKD', str(s)).encode('ascii', 'ignore').decode()
    return ' '.join(s.upper().split())

ETIQUETAS = {
    'TRAMO DE VIA': 'Tramo de vía', 'INTERSECCION': 'Intersección',
    'TUNEL': 'Túnel', 'VIA PEATONAL': 'Vía peatonal',
    'CICLORUTA': 'Ciclorruta', 'CICLORRUTA': 'Ciclorruta',
    'GLORIETA': 'Glorieta', 'PUENTE': 'Puente', 'PONTON': 'Pontón',
    'PASO A NIVEL': 'Paso a nivel', 'PASO ELEVADO': 'Paso elevado',
    'PASO INFERIOR': 'Paso inferior', 'LOTE O PREDIO': 'Lote o predio',
}

def etiqueta(s):
    k = _clave(s)
    if k not in ETIQUETAS:
        print(f'Categoría sin mapeo (revisar): {s!r}')
        return str(s).strip().capitalize()
    return ETIQUETAS[k]

# ---------------------------------------------------------
# Parámetros de la agrupación
# ---------------------------------------------------------
N_MINORITARIAS = 9          # cuántas categorías, las menos frecuentes, van a «Otras»
NOMBRE_OTRAS = 'Otras'
OTRAS_MANUAL = None         # o una lista de etiquetas, p. ej. ['Túnel', 'Pontón', ...]; si se da, anula N_MINORITARIAS

# ---------------------------------------------------------
# Conteos
# ---------------------------------------------------------
serie = df['DISENO_LUGAR_DESC'].dropna()
mapa = {v: etiqueta(v) for v in serie.unique()}              # se etiqueta una vez por valor único
conteos = serie.map(mapa).value_counts()                     # 'Cicloruta' y 'Ciclorruta' quedan juntas

n_total = int(conteos.sum())
n_sin_dato = int(df['DISENO_LUGAR_DESC'].isna().sum())
print(f'Categorías encontradas: {len(conteos)} | registros con dato: {n_total:,} | sin dato: {n_sin_dato:,}')

if OTRAS_MANUAL is not None:
    faltan = [c for c in OTRAS_MANUAL if c not in conteos.index]
    assert not faltan, f'Categorías de OTRAS_MANUAL que no existen: {faltan}'
    minoritarias = conteos.loc[OTRAS_MANUAL]
else:
    assert len(conteos) > N_MINORITARIAS, f'Hay solo {len(conteos)} categorías; no se pueden agrupar {N_MINORITARIAS}'
    orden_asc = conteos.sort_values(ascending=True, kind='stable')
    if orden_asc.iloc[N_MINORITARIAS - 1] == orden_asc.iloc[N_MINORITARIAS]:
        print('Aviso: empate en el corte entre «Otras» y las principales; revisar con OTRAS_MANUAL')
    minoritarias = orden_asc.iloc[:N_MINORITARIAS]

principales = conteos.drop(minoritarias.index).sort_values(ascending=False)
n_otras = int(minoritarias.sum())
final = pd.concat([principales, pd.Series({NOMBRE_OTRAS: n_otras})])    # «Otras» siempre al final

# Tabla de control
tabla = pd.DataFrame({'n': final, '%': (100 * final / n_total).round(1)})
print(tabla.to_string())
print('\nCategorías agrupadas en «Otras»:')
print(minoritarias.sort_values(ascending=False).to_frame('n').assign(**{'%': (100 * minoritarias / n_total).round(2)}).to_string())

# ---------------------------------------------------------
# Figura
# ---------------------------------------------------------
colores = ['#1F4E5F'] * len(principales) + ['#1F4E5F']        # «Otras» en gris: es un agregado
fig, ax = plt.subplots(figsize=(12, 6))
barras = ax.barh(final.index, final.values, color=colores, alpha=0.9)
ax.bar_label(barras, labels=[f'{miles(v)} ({pct_es(v / n_total)} %)' for v in final.values],
             padding=5, fontsize=9.5, fontweight='bold')

# Sin título interno: va en el pie de figura del documento
ax.set_xlabel('Número de siniestros', fontsize=10)     # si cada fila es un peatón afectado, rotular «Peatones afectados»
ax.set_ylabel('')
ax.invert_yaxis()
ax.set_xlim(0, final.max() * 1.22)
ax.xaxis.set_major_formatter(FuncFormatter(miles))
ax.spines['top'].set_visible(False); ax.spines['right'].set_visible(False)
ax.grid(axis='x', linestyle='--', linewidth=0.5, alpha=0.4)

plt.tight_layout()
plt.show()
```

### 6.5 Asociación entre variables categóricas y `ESTADO` (Chi-cuadrado y Cramér's V)

Antes de que Boruta decida por su cuenta dentro del CV espacial, conviene ver qué categóricas muestran asociación con el desenlace del peatón. Se usa Cramér's V (no solo el p-valor) porque con este volumen de datos casi cualquier asociación sale "significativa"; V mide qué tan fuerte es esa asociación, no solo si existe. **Nota:** el código usa `df['ESTADO']`, no `GRAVEDAD` — ver advertencia de la sección 6.0 sobre qué variable objetivo es la relevante para el modelo final.

```python
df.columns
```

```python
def cramers_v(tabla):
    chi2 = chi2_contingency(tabla)[0]
    n = tabla.sum().sum()
    r, k = tabla.shape
    return np.sqrt(chi2 / (n * (min(r, k) - 1)))

candidatas_categoricas = [
 'CONDICION', 'SEXO',
    'LOCALIDAD_DESC', 'DISENO_LUGAR_DESC', 'CLASE',
    'DIA_SEMANA', 'MES', 'HORA_DIA',
]

resultados = []
for col in candidatas_categoricas:
    if col not in df.columns:
        continue
    tabla = pd.crosstab(df[col], df['ESTADO'])
    tabla = tabla.loc[tabla.sum(axis=1) >= 5]  # descartar categorías con muy pocos casos
    if tabla.shape[0] < 2:
        continue
    chi2, p, dof, _ = chi2_contingency(tabla)
    v = cramers_v(tabla)
    resultados.append({'variable': col, 'chi2': chi2, 'p_valor': p, 'cramers_v': v, 'n_categorias': tabla.shape[0]})

tabla_bivariado_cat = pd.DataFrame(resultados).sort_values('cramers_v', ascending=False)
tabla_bivariado_cat
```

Como referencia orientativa para Cramér's V con 2-3 grados de libertad: <0.10 asociación despreciable, 0.10-0.30 débil, 0.30-0.50 moderada, >0.50 fuerte. Las variables con V muy bajo son candidatas a descartarse antes incluso de Boruta; ninguna decisión aquí es definitiva sin correrlo sobre los datos reales.


### 6.6 Asociación entre variables continuas y `ESTADO` (Kruskal-Wallis)

Se usa Kruskal-Wallis (no ANOVA) porque no hay razón para asumir normalidad ni varianzas iguales entre los grupos de `ESTADO`. Esta prueba es además la que debe decidir empíricamente si `EDAD` (o alguna variable derivada de vulnerabilidad etaria) es la que más separa a los grupos, en vez de asumirlo.

```python
candidatas_continuas = ['EDAD', 'HORA_DIA']

for col in candidatas_continuas:
    # dropna por grupo: kruskal no tolera NaN dentro de los arrays de entrada
    grupos = [g.dropna().values for _, g in df.groupby('ESTADO')[col]]
    grupos = [g for g in grupos if len(g) > 0]
    if len(grupos) < 2:
        continue
    stat, p = kruskal(*grupos)
    print(f"{col}: H={stat:.2f}, p-valor={p:.4g}")

fig, axes = plt.subplots(1, len(candidatas_continuas), figsize=(6*len(candidatas_continuas), 5))
for ax, col in zip(np.atleast_1d(axes), candidatas_continuas):
    sns.boxplot(data=df, x='ESTADO', y=col, ax=ax)  # ESTADO, no GRAVEDAD: ver nota abajo
    ax.set_title(f'{col} por estado')
plt.tight_layout()
plt.show()
```

## 8. Resumen y hallazgos clave



### 8.1 Hallazgos cuantitativos (con base en las salidas reales de este cuaderno)

- El subconjunto de peatones tiene **23.447 registros** a nivel de actor individual, correspondientes a **21.654 siniestros distintos** (10,9% del total de 199.146 siniestros de la base histórica) — confirmado en la sección 3.4.
- **1.793 siniestros** tienen más de un peatón involucrado (registros duplicados por `CODIGO_ACCIDENTE` dentro del subconjunto peatonal, sección 4.3). 
- Distribución de `ESTADO` (sección 6.0, sobre el subconjunto peatonal): `HERIDO` 22.012 (93,9%), `MUERTO` 1.433 (6,1%), `ILESO` 2 (0,01%) — confirma empíricamente el desbalance de clases.
- De las variables categóricas probadas contra `ESTADO` (sección 6.5), solo `SEXO` muestra una asociación que supera el umbral de "débil" (Cramér's V=0,113); `DISENO_LUGAR_DESC` (V=0,066), `HORA_DIA` (V=0,065), `LOCALIDAD_DESC` (V=0,033) y `CLASE` (V=0,031) son estadísticamente significativas pero con una fuerza de asociación despreciable dado el tamaño de muestra; `MES` y `DIA_SEMANA` ni siquiera alcanzan significancia (p=0,21 y p=0,64).
- De las variables continuas (sección 6.6), `EDAD` muestra una asociación fuerte y estadísticamente robusta con `ESTADO` (Kruskal-Wallis H=435,5, p≈2,7×10⁻⁹⁵); `HORA_DIA` no muestra ninguna asociación (H=0,34, p=0,85).
- `EDAD` y `HORA_DIA` no presentan multicolinealidad relevante entre sí (VIF≈2,8 para ambas, sección 6.7).
- 6,63% de los registros del subconjunto peatonal carecen de coordenadas y de variables derivadas del `FECHA_HORA_ACC` (concentrados en 2021, por el corte de cobertura de la hoja `SINIESTROS` documentado en la sección 5).


### 8.3 Conclusión general

Con la evidencia que este cuaderno produce realmente (no la que los encabezados originales sugerían), la única variable con señal fuerte sobre el desenlace del peatón es `EDAD`; el resto de las categóricas y `HORA_DIA` tienen una asociación estadísticamente detectable pero prácticamente irrelevante dado el tamaño de muestra. Esto **no** implica que esas variables deban descartarse del modelo de segmento — implica que a nivel de persona/accidente no explican mucho el desenlace individual, lo cual es consistente con la hipótesis de que el riesgo real vive en las características del segmento vial (diseño, velocidad regulada, exposición), no en quién es el peatón o cuándo ocurrió el choque. Antes de que estos resultados alimenten cualquier decisión de selección de variables del capítulo de metodología, los puntos 1 y 2 de la sección 8.2 tienen que resolverse — de lo contrario se estaría justificando la arquitectura del modelo final con un análisis exploratorio corrido sobre la variable y el nivel de análisis equivocados.
